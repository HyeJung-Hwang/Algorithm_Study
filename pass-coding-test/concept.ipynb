```python
# 람다 사용
# 1) 변수로서 참조 (간단 함수 정의)
add = lambda x,y: x+y
print(add(5,4))
# 2) 인수로 람다식 넘기기
num = [1,2,3,4,5]
squares = list(map(lambda x: x**2,num)) # list는 함수라서, iterable 넣으면 리스트로 변환
print(squares )
```

```python
# 조기 반환
def total_price(quantity,price):
    total = quantity * price
    if total > 100:
        return total*0.9
    return total
```
